# Anomalous protein–protein interactions in multivalent salt solution

*Coralie Pasquier, Mario Vazdar, Jan Forsman, Pavel Jungwirth, and Mikael Lund,
J. Phys. Chem. B 121, 3000–3006 (2017), [doi:10.1021/acs.jpcb.7b01051](https://doi.org/10.1021/acs.jpcb.7b01051)*

Two human serum albumin (HSA) molecules move along the axis of a periodic cylinder with explicit
trivalent yttrium ions, which move freely. Sites interact through a combined Debye–Hückel/Lennard-Jones
potential, and the protein–protein distance distribution gives the potential of mean force (PMF).

## 02 – Potentials of mean force

Plots the PMFs in `data/processed/pmf.csv` (from `01-simulate.ipynb`, implicit 1:1 salt, specific
Y³⁺–carboxylate attraction) and writes the figures to `figures/`. Faint lines are the raw PMFs;
solid lines are smoothing splines.


```python
from pathlib import Path

import matplotlib.pyplot as plt

from analysis.pmf import load_pmf, plot_pmf

ROOT = Path.cwd().parent
plt.rcParams.update({"font.size": 16, "figure.figsize": [7.0, 5.0]})
curves = load_pmf(ROOT / "data" / "processed" / "pmf.csv")
```

```python
for salt, title in [
    (0.0, "No 1:1 salt"),
    (0.1, "100 mM 1:1 salt"),
    (0.2, "200 mM 1:1 salt"),
]:
    fig, ax = plt.subplots()
    plot_pmf(ax, curves, salt)
    ax.set_title(title, verticalalignment="bottom")
    fig.savefig(ROOT / "figures" / f"pmf-{salt * 1000:.0f}mM.pdf", bbox_inches="tight")
    plt.show()
```
